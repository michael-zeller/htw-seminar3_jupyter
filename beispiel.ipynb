```python
# Imports
import matplotlib.pyplot as plt

# Daten: Tagestemperaturen einer Woche in °C
tage = ['Mo', 'Di', 'Mi', 'Do', 'Fr', 'Sa', 'So']
temperaturen = [12.5, 14.0, 11.8, 16.2, 18.5, 17.1, 13.9]

# Auswertung
minimum = min(temperaturen)
maximum = max(temperaturen)
durchschnitt = sum(temperaturen) / len(temperaturen)

print(f'Minimum:       {minimum:.1f} °C')
print(f'Maximum:       {maximum:.1f} °C')
print(f'Durchschnitt:  {durchschnitt:.1f} °C')
```

```python
# Visualisierung: das Diagramm erscheint direkt unter der Zelle
plt.figure(figsize=(7, 4))
plt.plot(tage, temperaturen, marker='o')
plt.axhline(durchschnitt, color='gray', linestyle='--', label='Durchschnitt')
plt.xlabel('Wochentag')
plt.ylabel('Temperatur (°C)')
plt.title('Temperaturen einer Woche')
plt.legend()
plt.grid(True)
plt.savefig('temperaturen.png')  # optional: Grafik speichern
plt.show()
```
